```python
from langgraph.graph import StateGraph, START,END
from langchain_openai import ChatOpenAI
from typing import TypedDict
from dotenv import load_dotenv

```

```python
load_dotenv()
llm=ChatOpenAI(
)
```

```python
class statellm():
    title: str
    outline: str
    content: str
    score: int
```

```python
def gen_outline(state:statellm)->statellm:
    title=state['title']
    prompt=f"Generate a detailed outline for a blog on the topic - {title}"
    outline=llm.invoke(prompt)
    
    state['outline']=outline
    
    return state
```

```python
def gen_blog(state:statellm)->statellm:
    title=state['title']
    outline=state['outline']
    prompt=f'Write a detailed blog on the title - {title} using the follwing outline \n {outline}'
    
    blog=llm.invoke(prompt)
    state['content']=blog
    
    return state
```

```python
def evaluation(state: statellm)->statellm:
    title=state['title']
    outline=state['outline']
    blog=state['content']
    
    prompt=f"evaluate the blog-{blog} based on the title-{title} and outline-{outline} out of 10"
    score=llm.invoke(prompt)
    
    state['score']=score
    
    return state
    
    
```

```python
graph=StateGraph(statellm)
graph.add_node("gen_outline",gen_outline)
graph.add_node("gen_blog",gen_blog)
graph.add_node("evaluation",evaluation)from frpofrofrfrofromgvfgchbvbhnjmfrrfrfrfrffr

graph.add_edge(START,'gen_outline')
graph.add_edge('gen_outline','gen_blog')
graph.add_edge('gen_blog','evaluation')
graph.add_edge('evaluation',END)

workflow = graph.compile()
```

```python
initial_state={
    'title': 'Rise of AI in Bangladesh'
}
final_state=workflow.invoke(initial_state)
print(final_state)
```
